```python
점수 = int(input("점수를 입력하세요 (0~100): "))

if 점수.isdigit():
    점수 = int(점수)
    
    if 점수 >= 90:
        print("A 학점입니다.")
    elif 점수 >= 80:
        print("B 학점입니다.")
    elif 점수 >= 70:
        print("C 학점입니다.")
    elif 점수 >= 60:
        print("D 학점입니다.")
    else:
        print("F 학점입니다.")

else:
    print('숫자를 입력하세요')
```

```python
coffee = 10
money = 300

while money:
    print(f'돈을 받았으니 커피를 줍니다. 커피가 {coffee}잔 남았습니다.')
    coffee -= 1
    if coffee == 0:
        print('커피가 떨어졌습니다. 영업 종료')
        break
```
